# Business Insights


The dataset contains 1,470 employees. Out of these, 237 employees left the organization, resulting in an overall attrition rate of 16.12%. The remaining 1,233 employees were retained, representing 83.88%.

Department-wise Attrition

Sales department has the highest attrition rate of 20.63%, followed by HR at 19.05%. The Research & Development department has the lowest attrition rate of 13.84%. This indicates that employee attrition is comparatively higher in the Sales department.

Job Role-wise Attrition

The Sales Representative role has the highest attrition rate of 39.76%, while the Research Director role has the lowest attrition rate of 2.50%. This indicates that attrition varies significantly across different job roles.

Age Group-wise Attrition

The 18–25 age group has the highest attrition rate of 35.77%, while the 36–45 age group has the lowest attrition rate of 9.19%. This indicates that younger employees are leaving the organization at a higher rate.

Gender-wise Attrition

The male employees have an attrition rate of 17.01%, while female employees have an attrition rate of 14.80%. The difference is relatively small, indicating that attrition rates are slightly higher among male employees in this dataset.

Overtime-wise Attrition

Employees who work overtime have an attrition rate of 30.53%, compared to 10.44% among employees who do not work overtime. This indicates a strong association between overtime and employee attrition.

Job Satisfaction-wise Attrition

Employees with Job Satisfaction Level 1 have the highest attrition rate, while employees with Level 4 have the lowest attrition rate. This indicates that lower job satisfaction is associated with higher employee attrition.

Work-Life Balance-wise Attrition

Employees with Work-Life Balance Level 1 have the highest attrition rate, while Level 3 employees have the lowest attrition rate. This suggests that poor work-life balance is associated with higher employee attrition.

Environment Satisfaction-wise Attrition

Employees with lower Environment Satisfaction levels show comparatively higher attrition rates, while employees with higher satisfaction levels show comparatively lower attrition rates.

Relationship Satisfaction-wise Attrition

Employee attrition rates vary across different Relationship Satisfaction levels. This suggests that employee relationships and workplace interactions may be associated with employee retention.

Job Involvement-wise Attrition

Employee attrition rates vary across different Job Involvement levels. This indicates that job involvement may be associated with employee retention. HR should examine employee engagement and involvement to identify possible areas for improvement.

Business Travel-wise Attrition

Employee attrition rates differ across business travel categories. This suggests that frequent business travel may be associated with employee attrition. HR should review travel requirements and employee workload.

Years at Company-wise Attrition

Employee attrition rates vary according to their years of service in the organization. This indicates that employee experience and length of service may be associated with employee retention.

Years Since Last Promotion-wise Attrition

Employee attrition rates vary according to the number of years since their last promotion. This suggests that promotion history may be associated with employee retention. HR should review career growth and promotion opportunities.

Monthly Income-wise Attrition

Employee attrition rates vary across different monthly income groups. This suggests that salary levels may be associated with employee retention. HR should review compensation structures to support employee satisfaction and retention.

# HR Recommendations

1. Improve Sales Employee Retention

The HR department should identify the reasons behind higher attrition in the Sales department. Regular feedback sessions, employee support and better work conditions can help improve employee retention.


2. Support Young Employees

The organization should provide proper training, mentorship programs and career development opportunities to young employees. This can help improve employee engagement and retention.

3. Manage Employee Overtime

The organization should monitor employee overtime and distribute workloads effectively. Providing flexible working hours and sufficient rest can help reduce employee stress and improve retention.

4. Improve Job Satisfaction

The organization should conduct regular employee feedback surveys, understand employee concerns and improve workplace policies. Recognizing employee performance can also help improve job satisfaction and retention.

5. Improve Work-Life Balance

The organization should promote a healthy work-life balance by providing flexible schedules, managing workloads and encouraging employees to maintain a balance between professional and personal responsibilities.

6. Provide Career Growth Opportunities

The organization should provide regular training, skill development programs and fair promotion opportunities. Clear career growth paths can help employees stay motivated and improve long-term retention.

7. Strengthen Workplace Relationships

The organization should encourage teamwork, communication and collaboration among employees. Regular team activities and supportive management can help create a positive working environment.

8. Review Business Travel Requirements

The organization should review employee travel schedules and avoid unnecessary business trips. Proper travel planning and flexible arrangements can help improve employee comfort and satisfaction.

9. Review Employee Compensation

The organization should regularly review salary structures and ensure fair compensation based on employee roles, skills and responsibilities. Competitive compensation can support employee satisfaction and retention.

10. Improve Employee Engagement

The organization should organize employee engagement activities, conduct regular performance discussions and encourage employee participation in decision-making. This can help build a positive workplace culture and improve employee retention.
