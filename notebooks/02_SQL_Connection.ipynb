```python
import mysql.connector
import pandas as pd
```

```python
import os

connection = mysql.connector.connect(
    host="localhost",
    user="root",
    password=os.getenv("MYSQL_PASSWORD"),
    database="peoplepulse"
)
```

# Total Employee Analysis

```python
cursor = connection.cursor()

cursor.execute("SELECT COUNT(*) FROM employee_attrition")

result = cursor.fetchone()

print("Total Employees:", result[0])
```

### Business Insight

The dataset contains 1470 employees.

This analysis helps us understand the total number of employees in the organization.

# Department-wise Attrition Analysis

```python
query = """
SELECT Department, Attrition, COUNT(*) AS Employee_Count
FROM employee_attrition
GROUP BY Department, Attrition
ORDER BY Department;
"""

cursor.execute(query)

results = cursor.fetchall()

df_sql = pd.DataFrame(
    results,
    columns=["Department", "Attrition", "Employee_Count"]
)

display(df_sql)
```

```python
query = """
SELECT Gender, Attrition, COUNT(*) AS Employee_Count
FROM employee_attrition
GROUP BY Gender, Attrition;
"""

cursor.execute(query)
results = cursor.fetchall()

df_gender = pd.DataFrame(
    results,
    columns=["Gender", "Attrition", "Employee_Count"]
)

display(df_gender)
```

### Business Insight

- Sales department has the highest number of employees leaving (92).
- Research & Development has 133 employees leaving.
- Human Resources has the lowest number of employees leaving (12).

This analysis helps identify employee attrition across different departments.

# Gender-wise Attrition Analysis

```python
query = """
SELECT Gender, Attrition, COUNT(*) AS Employee_Count
FROM employee_attrition
GROUP BY Gender, Attrition;
"""

cursor.execute(query)
results = cursor.fetchall()

df_gender = pd.DataFrame(
    results,
    columns=["Gender", "Attrition", "Employee_Count"]
)

display(df_gender)
```

### Business Insight

This analysis compares employee attrition between male and female employees.

It helps understand the gender-wise distribution of employee attrition.

# Overtime Analysis

```python
query = """
SELECT OverTime, COUNT(*) AS Employee_Count
FROM employee_attrition
GROUP BY OverTime;
"""

cursor.execute(query)
results = cursor.fetchall()

df_overtime = pd.DataFrame(
    results,
    columns=["OverTime", "Employee_Count"]
)

display(df_overtime)
```

### Business Insight

This analysis shows how many employees work overtime and how many do not.

It helps understand the distribution of overtime among employees.

# Average Monthly Income by Department

```python
query = """
SELECT Department, AVG(MonthlyIncome) AS Average_Income
FROM employee_attrition
GROUP BY Department;
"""

cursor.execute(query)
results = cursor.fetchall()

df_income = pd.DataFrame(
    results,
    columns=["Department", "Average_Income"]
)

display(df_income)
```

### Business Insight

This analysis compares the average monthly income across different departments.

It helps understand income differences between departments.
